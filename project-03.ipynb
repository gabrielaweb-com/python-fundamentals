```
url = "https://sistema.exemplo.com/busca?termo=velho"

url_atualizada = url.replace("velho", "novo")

def extrair_termo(url):
    if "termo=" in url:
        inicio = url.find("termo=") + len("termo=")

        fim = url.find("&", inicio)

        if fim == -1:
            return url[inicio:]
        else:
            return url[inicio:fim]

    return None

termo_extraido = extrair_termo(url_atualizada)
 
print(url_atualizada)
print(termo_extraido)
```

```
def criptografar_cme(lista_pedidos, status="Pendente"):
    for codigo in lista_pedidos:

        mensagem = f"Pedido {codigo}: Status - {status}"

        mensagem_criptografada = ""

        for indice, caractere in enumerate(mensagem):

            if caractere == " ":
                mensagem_criptografada += " "

            elif indice % 2 == 0:
                mensagem_criptografada += chr(ord(caractere) + 2)

            else:
                mensagem_criptografada += chr(ord(caractere) - 1)

        print(mensagem_criptografada)

pedidos = ["A123", "B456", "C789"]

criptografar_cme(pedidos)

criptografar_cme(pedidos, status="Enviado")
```

```
def ordenar_valores(valores):

    for i in range(len(valores)):

        for j in range(i + 1, len(valores)):

            if valores[i] > valores [j]:
                valores[i], valores[j] = valores[j], valores[i]

    return valores

numeros = [8, 3, 1, 7, 4, 2]

resultado = ordenar_valores(numeros)

print(resultado)
```

```
def mover_hanoi(n, origem, destino, suporte):
    """
    Resolve o problema da Torre de Hanói utilizando recursão.
    Parâmetros:
    n -> quantidade de discos
    origem -> torre inicial
    destino -> torre final
    suporte -> torre auxiliar
    A função apenas imprime os movimentos dos discos.
    """

    if n == 1:
        print(f"Mover disco de {origem} para { destino}")

    else:
        mover_hanoi(n - 1, origem, suporte, destino)

        print(f"Mover disco de {origem} para {destino}")

        mover_hanoi(n - 1, suporte, destino, origem)
        
mover_hanoi(3, "A", "C", "B")
```

```
def extrair_codigos(mensagem):

    codigos = [] 
    i=0

    while i < len(mensagem):

        if mensagem[i] == "@":

            inicio = i + 1
            fim = inicio

            while fim < len(mensagem) and mensagem[fim] != "#":
                fim += 1

            trecho = mensagem[inicio:fim]

            codigo_limpo = ""

            for caractere in trecho:
                if caractere.isalnum():
                    codigo_limpo += caractere

            codigos.append(codigo_limpo)

            i = fim 
        i += 1

    return codigos

mensagem = "Sensor detectado: @A1B2C3# fora de faixa. Erro: @9X # ignorado.\nValidação: @99Z#"

codigos = extrair_codigos(mensagem)

print(codigos)
```

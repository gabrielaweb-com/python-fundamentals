```
preco_unitario = 12.50
quantidade = 47 
percentual_desconto = 10
percentual_imposto = 15
tamanho_lote = 12

def gerar_relatorio(preco_unitario, quantidade, percentual_desconto, percentual_imposto, tamanho_lote):
    valor_bruto = preco_unitario * quantidade
    valor_desconto = valor_bruto * percentual_desconto / 100
    base_tributavel = valor_bruto - valor_desconto 
    valor_imposto = base_tributavel * percentual_imposto / 100
    total_liquido = base_tributavel + valor_imposto
    lotes_fechados = quantidade // tamanho_lote
    sobras = quantidade % tamanho_lote

    print("### Relatório de Cálculos de Precificação")
    print("| Etapa             | Resultado |")
    print("|-------------------|-----------|")
    print(f"| Valor Bruto       | {valor_bruto:.2f} |")
    print(f"| Valor do Desconto | {valor_desconto:.2f} |")
    print(f"| Base Tributável   | {base_tributavel:.2f} |")
    print(f"| Valor do Imposto  | {valor_imposto:.2f} |")
    print(f"| Total Líquido     | {total_liquido:.2f} |")
    print(f"| Lotes Fechados    | {lotes_fechados} |")
    print(f"| Sobras            | {sobras} |")

gerar_relatorio(
    preco_unitario,
    quantidade,
    percentual_desconto,
    percentual_imposto,
    tamanho_lote
)
```

```
def calcular_media(notas, pesos):
    soma_ponderada = sum(nota * peso for nota, peso in zip(notas, pesos))
    soma_pesos = sum(pesos)
    return soma_ponderada / soma_pesos

def arredondar_media(media):
    return round(media, 1)

def classificar_desempenho(media):
    if media < 2.5:
        return "ND"
    elif media < 5.0:
        return "D"
    elif media < 7.5:
        return "DL"
    else:
        return "DML"

notas = [8.0, 6.5, 9.0]
pesos = [2, 3, 5]

media_calculada = calcular_media(notas,pesos)
media_final = arredondar_media(media_calculada)
classificacao_final = classificar_desempenho(media_final)

print("Notas:", notas)
print("Pesos:", pesos)
print()
print(f"Média calculada: {media_calculada:.2f}")
print(f"Média arredondada: {media_final}")
print(f"Classificação final: {classificacao_final}")

print("\n--------------------\n")

notas = [4.0, 5.0, 6.0]
pesos = [1, 2, 3]

media_calculada = calcular_media(notas, pesos)
media_final = arredondar_media(media_calculada)
classificacao_final = classificar_desempenho(media_final)

print("Notas:", notas)
print("Pesos:", pesos)
print()
print(f"Média calculada: {media_calculada:.2f}")
print(f"Média arredondada: {media_final}")
print(f"Classificação final: {classificacao_final}")
```

```
def calcular_distancia(passos, tamanho_passo):
    """
    Calcula a distância total percorrida em um armazém. 

Args: 
    passos (list of int): Uma lista de valores inteiros representando o número de passos em cada deslocamento. 
    tamanho_passo (float): O comprimento de cada passo em metros. 

Returns: 
    float: A distância total percorrida em metros. 

Example: 
    >>> calcular_distancia([10, 20, 15], 0.8) 
    36.0 
    """
    return sum(passos) * tamanho_passo

print(calcular_distancia([10, 20, 15], 0.8))
print(calcular_distancia([5, 10, 5], 0.7))

def calcular_velocidade(distancia, tempo, unidade="m/s"):
    """
    Calcula a velocidade de um deslocamento a partir da distância percorrida e do tempo utilizado.

    Args:
        distancia (float): Valor da distância percorrida.
        tempo (float): Tempo gasto para percorrer a distância.
        unidade (str): Unidade em que a velocidade será retornada.
            As opções disponíveis são "m/s", "km/h" e "mph".
            O valor padrão é "m/s".
        
    Returns:
        float: Valor da velocidade calculada na unidade especificada.
        None: Retornado quando o tempo informado é menor ou igual a zero.
    
    Example:
        >>> calcular_velocidade(50, 10)
        5.0

        >>> calcular_velocidade(50, 10, "km/h")
        18.0
    """

    if tempo <= 0:
        return None
    
    velocidade = distancia / tempo

    if unidade == "km/h":
        velocidade *= 3.6
    elif unidade == "mph":
        velocidade *= 2.23694
    
    return velocidade

print(calcular_velocidade(100, 20))
print(calcular_velocidade(100, 20, "km/h"))
print(calcular_velocidade(100, 20, "mph"))

help(calcular_distancia)
help(calcular_velocidade)
```

```
def janelas_palindromas(s, k):
    resultado = []

    for i in range(len(s) - k + 1):
        janela = s[i:i + k]

        if janela == janela[::-1]:
            resultado.append(janela)

    return resultado

print(janelas_palindromas("banana", 3))

def janelas_palindromas_oneliner(s, k):
    return [s[i:i + k] for i in range(len(s) - k + 1) if s[i:i + k] == s[i:i + k] [::-1]]

print(janelas_palindromas_oneliner("banana", 3))
```
